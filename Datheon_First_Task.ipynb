```python
#3 Extract Year-Month for seasonal analysis (e.g '2023-01')
df['Year-Month'] = df['Order Date'].dt.to_period('M')

print("Year-Month column added successfully.")
print("\nData types and shape after cleaning:")
print(df.info())
```

```python
#4 Check for duplicates
duplicates = df.duplicated()
print("Number of duplicate rows:", duplicates.sum())
```

```python
#5 Remove duplicates if any
if duplicates.sum() > 0:
    df = df.drop_duplicates()
    print("Duplicates removed. New shape:", df.shape)
```

```python
# Group by category
category_perf = df.groupby('Category')[['Sales', 'Profit']].sum().reset_index()
print("--- Performance by Category ---")
display(category_perf)
```

```python
# Group by Region
region_perf = df.groupby('Region')[['Sales', 'Profit']].sum().reset_index()
print("\n--- Performance by region ---")
display(region_perf)
```

```python
# Group by Customer Segment
segment_perf = df.groupby('Segment')[['Sales', 'Profit']].sum().reset_index
print("\n--- Performance by Customer Segment ---")
display(segment_perf)
```
